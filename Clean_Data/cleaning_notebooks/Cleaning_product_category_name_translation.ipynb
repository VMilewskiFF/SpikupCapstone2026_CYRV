# Cleaning_product_category_name_translation

## Conclusion

The cleaned product category translation dataset contains 73 unique Portuguese-to-English category mappings and provides complete coverage of all non-null product categories in the cleaned products dataset.

The original 71 mappings were retained without modification. Two additional mappings were added for categories that existed in the raw and cleaned products datasets but were missing from the original translation table. These additions were documented explicitly to distinguish them from the original source mappings.

The dataset is now ready to be saved and used as a complete product category lookup table for downstream analysis, joins, and visualizations.

## Cleaning Summary

The product category translation dataset was inspected and validated as a lookup table between Portuguese and English product category names.

### Cleaning actions performed

- Verified that the original dataset contained 71 unique Portuguese-to-English category mappings.
- Confirmed that there were no missing values or duplicated rows.
- Confirmed that both Portuguese and English category names were unique.
- Checked category names for leading/trailing spaces, uppercase characters, empty values, and whitespace formatting issues; no formatting problems were found.
- Compared the translation table with the cleaned products dataset.
- Identified two product categories that were present in the products dataset but missing from the original translation table:
  - `pc_gamer`
  - `portateis_cozinha_e_preparadores_de_alimentos`
- Confirmed that these categories were already present in the raw products dataset and were not introduced during data cleaning.
- Added two mappings to provide complete translation coverage for downstream analysis:
  - `pc_gamer` → `pc_gamer`
  - `portateis_cozinha_e_preparadores_de_alimentos` → `portable_kitchen_food_preparers`
- The two additional mappings were added specifically to complete the lookup table and were not present in the original translation dataset.

### Final validation

- Final shape: 73 rows × 2 columns
- Missing values: 0
- Fully duplicated rows: 0
- Duplicated Portuguese categories: 0
- Duplicated English categories: 0
- Product categories covered by the translation table: 73 / 73
- Missing product category translations: 0
- Extra translation categories not present in products: 0

```python
import pandas as pd

translation = pd.read_csv(
    "../../data/cyrv/product_category_name_translation.csv"
)

print("Dataset shape:", translation.shape)

translation.head()
```

```python
print("Data types:")
print(translation.dtypes)

print("\nMissing values:")
print(translation.isna().sum())
```

```python
print(
    "Duplicate rows:",
    translation.duplicated().sum()
)

print(
    "Duplicate Portuguese categories:",
    translation["product_category_name"].duplicated().sum()
)

print(
    "Duplicate English categories:",
    translation["product_category_name_english"].duplicated().sum()
)

print(
    "Unique Portuguese categories:",
    translation["product_category_name"].nunique()
)

print(
    "Unique English categories:",
    translation["product_category_name_english"].nunique()
)
```

```python
for column in [
    "product_category_name",
    "product_category_name_english"
]:
    print(f"\n{column}")

    print(
        "Leading/trailing spaces:",
        translation[column]
        .ne(translation[column].str.strip())
        .sum()
    )

    print(
        "Contains uppercase:",
        translation[column]
        .str.contains(r"[A-Z]", regex=True)
        .sum()
    )

    print(
        "Empty/whitespace-only:",
        translation[column]
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        "Contains spaces:",
        translation[column]
        .str.contains(r"\s", regex=True)
        .sum()
    )
```

```python
products = pd.read_csv(
    "../CYRV_products_cleaned.csv"
)

product_categories = set(
    products["product_category_name"]
    .dropna()
    .unique()
)

translation_categories = set(
    translation["product_category_name"]
    .unique()
)

missing_translations = sorted(
    product_categories - translation_categories
)

extra_translations = sorted(
    translation_categories - product_categories
)

print(
    "Unique categories in products:",
    len(product_categories)
)

print(
    "Categories in translation table:",
    len(translation_categories)
)

print(
    "\nMissing translations:",
    missing_translations
)

print(
    "\nExtra translations:",
    extra_translations
)
```

```python
print(
    products[
        products["product_category_name"].isin(
            missing_translations
        )
    ]["product_category_name"]
    .value_counts()
)
```

```python
from difflib import get_close_matches

for category in missing_translations:
    similar = get_close_matches(
        category,
        translation["product_category_name"].tolist(),
        n=5,
        cutoff=0.4
    )

    print(f"\nMissing category: {category}")
    print("Similar categories:", similar)
```

```python
products[
    products["product_category_name"].isin(
        missing_translations
    )
]
```

```python
raw_products = pd.read_csv(
    "../../data/cyrv/CYRV_products_dataset.csv"
)

print(
    raw_products[
        raw_products["product_category_name"].isin(
            missing_translations
        )
    ]["product_category_name"]
    .value_counts()
)
```

```python
pd.set_option("display.max_rows", None)

translation.sort_values(
    "product_category_name"
)
```

```python
additional_translations = pd.DataFrame({
    "product_category_name": [
        "pc_gamer",
        "portateis_cozinha_e_preparadores_de_alimentos"
    ],
    "product_category_name_english": [
        "pc_gamer",
        "portable_kitchen_food_preparers"
    ]
})

translation = pd.concat(
    [translation, additional_translations],
    ignore_index=True
)

print("Updated shape:", translation.shape)

translation.tail()
```

```python
translation_categories = set(
    translation["product_category_name"].unique()
)

missing_translations = sorted(
    product_categories - translation_categories
)

extra_translations = sorted(
    translation_categories - product_categories
)

print("Product categories:", len(product_categories))
print("Translation categories:", len(translation_categories))
print("Missing translations:", missing_translations)
print("Extra translations:", extra_translations)
```

```python
print("Final shape:", translation.shape)

print("\nMissing values:")
print(translation.isna().sum())

print(
    "\nDuplicate rows:",
    translation.duplicated().sum()
)

print(
    "Duplicate Portuguese categories:",
    translation["product_category_name"].duplicated().sum()
)

print(
    "Duplicate English categories:",
    translation["product_category_name_english"].duplicated().sum()
)
```

```python
output_path = "../product_category_name_translation_cleaned.csv"

translation.to_csv(
    output_path,
    index=False
)

print(f"Saved to: {output_path}")
```
